```
import matplotlib.pyplot as plt
days = [1, 2, 3, 4, 5]
temperature = [30, 32, 31, 35, 34]
plt.plot(days, temperature, label='temperature',
    color='red',
    linestyle='--',
    linewidth=2,
    marker='o',
    markersize=8,
    markerfacecolor='yellow',
    markeredgecolor='black',
         alpha=0.8

)
plt.title('Daily Temparature')
plt.xlabel('Days')
plt.ylabel('Temparature')
plt.grid()
plt.legend()
plt.show()
```
