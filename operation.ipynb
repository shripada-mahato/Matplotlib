Days and temparature Analysis

```python
import matplotlib.pyplot as plt
days = [1, 2, 3, 4, 5]
temperature = [30, 32, 31, 35, 34]
plt.plot(days,temperature)
plt.title("Daily Tempareture")
plt.xlabel("Day")
plt.ylabel("Temperature")
plt.show()
```

grid() function

```python
import matplotlib.pyplot as plt
days = [1, 2, 3, 4, 5]
temperature = [30, 32, 31, 35, 34]
plt.grid(True)
plt.title("Daily Tempareture")
plt.xlabel("Day")
plt.ylabel("Temperature")
plt.show()
```
