Days and temparature Analysis

```python
import matplotlib.pyplot as plt
days = [1, 2, 3, 4, 5]
temperature = [30, 32, 31, 35, 34]
plt.plot(days,temperature)
plt.title("Daily Tempareture")
plt.xlabel("Day")
plt.ylabel("Temperature")
plt.show()
```

grid() function

```python
import matplotlib.pyplot as plt
days = [1, 2, 3, 4, 5]
temperature = [30, 32, 31, 35, 34]
plt.plot(days,temperature)
plt.grid(True)
plt.title("Daily Tempareture")
plt.xlabel("Day")
plt.ylabel("Temperature")
plt.show()
```

label() method

```python
import matplotlib.pyplot as plt
days = [1, 2, 3, 4, 5]
temperature = [30, 32, 31, 35, 34]
plt.plot(days,temperature, label='temperature')
plt.legend()
plt.grid(True)
plt.title("Daily Tempareture")
plt.xlabel("Day")
plt.ylabel("Temperature")
plt.show()
```

color() method

```python
import matplotlib.pyplot as plt
days = [1, 2, 3, 4, 5]
temperature = [30, 32, 31, 35, 34]
plt.plot(days,temperature, label='temperature',color='red')
plt.legend()
plt.grid(True)
plt.title("Daily Tempareture")
plt.xlabel("Day")
plt.ylabel("Temperature")
plt.show()
```

linestyle() method

```python
import matplotlib.pyplot as plt
days = [1, 2, 3, 4, 5]
temperature = [30, 32, 31, 35, 34]
plt.plot(days,temperature, 
         label='temperature',
         color='red',
         linestyle='--')
plt.legend()
plt.grid(True)
plt.title("Daily Tempareture")
plt.xlabel("Day")
plt.ylabel("Temperature")
plt.show()
```

linewidth() method

```python
import matplotlib.pyplot as plt
days = [1, 2, 3, 4, 5]
temperature = [30, 32, 31, 35, 34]
plt.plot(days,temperature, 
         label='temperature',
         color='red',
         linestyle='--',
         linewidth=3)
plt.legend()
plt.grid(True)
plt.title("Daily Tempareture")
plt.xlabel("Day")
plt.ylabel("Temperature")
plt.show()
```

```python
import matplotlib.pyplot as plt
days = [1, 2, 3, 4, 5]
temperature = [30, 32, 31, 35, 34]
plt.plot(
    days,
    temperature,
    label = 'temperature',
    color='y',
    linestyle='--',
    linewidth='3',
    
)
plt.title("practicing")
plt.xlabel('days')
plt.ylabel('temparature')
plt.grid()
plt.legend()
plt.show()
```

marker function

```python
import matplotlib.pyplot as plt
days = [1, 2, 3, 4, 5]
temperature = [30, 32, 31, 35, 34]
plt.plot(
    days,temperature, label = 'days',
    color='r',
    linestyle = '--',
    linewidth = 3,
    marker = 'o',
    markersize = 10
)
plt.title('temperature graph')
plt.legend()
plt.grid()
plt.xlabel('days')
plt.ylabel('temparature')
plt.show()
```

marker

```python
import matplotlib.pyplot as plt
days = [1, 2, 3, 4, 5]
temperature = [30, 32, 31, 35, 34]
plt.plot(
    days,temperature, label = 'temperature',
    color='r',
    linestyle = '--',
    linewidth = 2,
    marker = 'o',
    markersize = 10,
    markerfacecolor = 'yellow'
)
plt.title('temperature graph')
plt.legend()
plt.grid()
plt.xlabel('days')
plt.ylabel('temparature')
plt.show()
```

```python
import matplotlib.pyplot as plt
days = [1, 2, 3, 4, 5]
temperature = [30, 32, 31, 35, 34]
plt.plot(
    days,temperature, label='days',
    color = 'r',
    linestyle = '--',
    marker = 'o',
    markerfacecolor='yellow',
    markersize = 10,
    linewidth=3,
    markeredgecolor='black'

)
plt.title('temperature graph')
plt.xlabel('days')
plt.ylabel('temperature')
plt.legend()
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt
days = [1, 2, 3, 4, 5]
temperature = [30, 32, 31, 35, 34]
plt.plot(
    days,temperature, label='temperature',
    color = 'r',
    linestyle = '--',
    marker = 'o',
    markerfacecolor='yellow',
    markersize = 10,
    linewidth=3,
    markeredgecolor='black',
    markeredgewidth=2,
    alpha=0.7

)
plt.title('temperature graph')
plt.xlabel('days')
plt.ylabel('temperature')
plt.legend()
plt.grid()
plt.show()
```
