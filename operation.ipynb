Days and temparature Analysis

```python
import matplotlib.pyplot as plt
days = [1, 2, 3, 4, 5]
temperature = [30, 32, 31, 35, 34]
plt.plot(days,temperature)
plt.title("Daily Tempareture")
plt.xlabel("Day")
plt.ylabel("Temperature")
plt.show()
```

grid() function

```python
import matplotlib.pyplot as plt
days = [1, 2, 3, 4, 5]
temperature = [30, 32, 31, 35, 34]
plt.plot(days,temperature)
plt.grid(True)
plt.title("Daily Tempareture")
plt.xlabel("Day")
plt.ylabel("Temperature")
plt.show()
```

label() method

```python
import matplotlib.pyplot as plt
days = [1, 2, 3, 4, 5]
temperature = [30, 32, 31, 35, 34]
plt.plot(days,temperature, label='temperature')
plt.legend()
plt.grid(True)
plt.title("Daily Tempareture")
plt.xlabel("Day")
plt.ylabel("Temperature")
plt.show()
```

color() method

```python
import matplotlib.pyplot as plt
days = [1, 2, 3, 4, 5]
temperature = [30, 32, 31, 35, 34]
plt.plot(days,temperature, label='temperature',color='red')
plt.legend()
plt.grid(True)
plt.title("Daily Tempareture")
plt.xlabel("Day")
plt.ylabel("Temperature")
plt.show()
```

linestyle() method

```python
import matplotlib.pyplot as plt
days = [1, 2, 3, 4, 5]
temperature = [30, 32, 31, 35, 34]
plt.plot(days,temperature, 
         label='temperature',
         color='red',
         linestyle='--')
plt.legend()
plt.grid(True)
plt.title("Daily Tempareture")
plt.xlabel("Day")
plt.ylabel("Temperature")
plt.show()
```

linewidth() method

```python
import matplotlib.pyplot as plt
days = [1, 2, 3, 4, 5]
temperature = [30, 32, 31, 35, 34]
plt.plot(days,temperature, 
         label='temperature',
         color='red',
         linestyle='--',
         linewidth=3)
plt.legend()
plt.grid(True)
plt.title("Daily Tempareture")
plt.xlabel("Day")
plt.ylabel("Temperature")
plt.show()
```

marker function

```python
import matplotlib.pyplot as plt
days = [1, 2, 3, 4, 5]
temperature = [30, 32, 31, 35, 34]
plt.plot(
    days,temperature, label = 'days',
    color='r',
    linestyle = '--',
    linewidth = 3,
    marker = 'o',
    markersize = 10
)
plt.title('temperature graph')
plt.legend()
plt.grid()
plt.xlabel('days')
plt.ylabel('temparature')
plt.show()
```

```python
import matplotlib.pyplot as plt
hours = [1, 2, 3, 4, 5]
marks = [40, 50, 55, 65, 80]
plt.scatter(hours,marks,
            color = 'blue',
            marker='o',
            alpha=0.7,
            s=105)
plt.title('study hour vs marks')
plt.xlabel('study hour')
plt.ylabel('marks')
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt
year = [2021,2022,2023,2024,2025]
revenue = [15000000,12000000,9000000,23000000,18000000]
plt.figure(figsize=(20,5))


plt.subplot(1,4,1)
plt.plot(year,revenue)
plt.grid()
plt.title('Plot Chart')
plt.xlabel('Year')
plt.ylabel('Revenue')

plt.subplot(1,4,2)
plt.bar(year,revenue,color='green')
plt.title('Bar Chart')
plt.xlabel('Year')
plt.ylabel('Revenue')


plt.subplot(1,4,3)
plt.hist(year)
plt.title('Hist Chart')
plt.xlabel('Year')
plt.ylabel('Revenue')

plt.subplot(1,4,4)
plt.pie(revenue,frame=(1,2,3,4,5),shadow=10)
plt.title('Pie Chart')
plt.xlabel('Year')
plt.ylabel('Revenue')
plt.show()
```

```python
import matplotlib.pyplot as plt
days = [1, 2, 3, 4, 5]
sales = [20, 35, 30, 45, 50]

plt.bar(days, sales,
        color='green')
plt.xticks(days,["Mon", "Tue", "Wed", "Thu", "Fri"])
plt.xlabel('Days')
plt.ylabel('Sales')
plt.title('Weekly Sales')
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt
products = ["A", "B", "C", "D"]
sales = [20, 40, 60, 80]

plt.bar(products, sales,
        color='green')
plt.yticks([0, 20, 40, 60, 80, 100])
plt.xlabel('products')
plt.ylabel('Sales')
plt.title('product Sales')
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [20, 35, 30, 50, 80]
plt.plot(months,sales,
         marker='o',
         color='green',
         markerfacecolor='yellow')
plt.annotate('Higest Sales',xy=('May',80))
plt.xlabel('Month')
plt.ylabel('Sales')
plt.grid()
plt.show()
```
